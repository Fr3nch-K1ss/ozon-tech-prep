```python
import pandas as pd
import numpy as np
df = pd.read_csv("./data/data.csv", encoding='latin1')
print('Информация по таблице:')
print('Размер:', df.shape)
print()
print(df.describe())
print()
print(df.info())
```

```python
print('Первые 20 строк:')
print(df.head(20))


print('Средняя цена за товар',df["UnitPrice"].mean())
print('Медианная цена за товар',df["UnitPrice"].median())
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
print("Таблица с ячейкой выручки")
df.head()
```

```python
sales = df[df["Quantity"] > 0]
total_revenue = sales["Revenue"].sum()

print(f"Общая выручка: £{total_revenue:,.2f}")
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Month"] = df["InvoiceDate"].dt.to_period("M")

sales = df[df["Quantity"] > 0]
sales.groupby("Month")["Revenue"].sum()
```
